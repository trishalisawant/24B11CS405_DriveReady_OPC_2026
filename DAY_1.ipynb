````python

# ============================================================
# AI-POWERED SOFTWARE BUG DETECTION AGENT
# ============================================================

# 1. IMPORT LIBRARIES

import os
import ast
import json

from google.colab import userdata
from langchain_groq import ChatGroq

print("=" * 60)
print("AI SOFTWARE BUG DETECTION AGENT")
print("=" * 60)


# 2. LOAD GROQ API KEY

api_key = userdata.get("GROQ_API_KEY")

if not api_key:
    raise ValueError("Groq API key not found in Colab Secrets!")

os.environ["GROQ_API_KEY"] = api_key

print("\nGroq API key loaded successfully!")


# 3. INITIALIZE GROQ AI MODEL

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)

print("Groq AI model initialized successfully!")


# 4. TEST AI CONNECTION

response = llm.invoke(
    "What is a software bug? Explain in simple English."
)

print("\nAI CONNECTION TEST:")
print(response.content)


# 5. SYNTAX ERROR DETECTION

def detect_syntax_error(code):

    try:
        ast.parse(code)

        return {
            "status": "success",
            "message": "No syntax errors found."
        }

    except SyntaxError as error:

        return {
            "status": "error",
            "message": error.msg,
            "line": error.lineno
        }


# 6. BASIC CODE ANALYZER

def analyze_code(code):

    results = []

    try:
        tree = ast.parse(code)

    except SyntaxError as error:

        return [{
            "type": "Syntax Error",
            "message": error.msg,
            "line": error.lineno
        }]

    assigned = set()
    used = set()

    for node in ast.walk(tree):

        if isinstance(node, ast.Name):

            if isinstance(node.ctx, ast.Store):
                assigned.add(node.id)

            elif isinstance(node.ctx, ast.Load):
                used.add(node.id)

    for variable in assigned - used:

        results.append({
            "type": "Possible Unused Variable",
            "message": f"Variable '{variable}' may not be used."
        })

    if not results:

        results.append({
            "type": "Success",
            "message": "No basic issues detected."
        })

    return results


# 7. AI BUG EXPLANATION ENGINE

def explain_bug_with_ai(code, issues):

    prompt = f"""
    You are an expert Python software debugging assistant.

    Analyze the following Python code:

    {code}

    Detected issues:

    {issues}

    Your tasks:

    1. Explain each detected issue.
    2. Explain why it happened.
    3. Suggest how to fix it.
    4. Provide corrected Python code.
    5. Explain the corrected code in simple English.

    Do not invent errors.
    Clearly distinguish confirmed errors from possible warnings.
    """

    response = llm.invoke(prompt)

    return response.content


# 8. SAMPLE PYTHON CODE

sample_code = """
a = 10
b = 20
result = a + b
print(a)
"""


# 9. RUN BUG DETECTION

print("\n" + "=" * 60)
print("BUG DETECTION STARTED")
print("=" * 60)

print("\nSUBMITTED CODE:")
print(sample_code)

issues = analyze_code(sample_code)

print("\nDETECTED ISSUES:")

for issue in issues:
    print(issue)


# 10. GET AI EXPLANATION

print("\n" + "=" * 60)
print("AI BUG EXPLANATION")
print("=" * 60)

ai_explanation = explain_bug_with_ai(
    sample_code,
    issues
)

print(ai_explanation)


# 11. FINAL STATUS

print("\n" + "=" * 60)
print("BUG ANALYSIS COMPLETED")
print("=" * 60)

print("Syntax Analysis:", detect_syntax_error(sample_code))
print("AI Explanation: Generated successfully")

print("\nAI SOFTWARE BUG DETECTION AGENT EXECUTED!")
````

````python

# ============================================================
# PHASE 4: AUTOMATIC CODE FIXING ENGINE
# ============================================================

import re
import ast


# 1. EXTRACT PYTHON CODE FROM AI RESPONSE

def extract_code(ai_response):

    # Find Python code inside Markdown code blocks
    pattern = r"```(?:python)?\s*(.*?)```"

    matches = re.findall(
        pattern,
        ai_response,
        re.DOTALL | re.IGNORECASE
    )

    if matches:
        return matches[0].strip()

    # If no code block is found, return the response
    return ai_response.strip()


# 2. AUTOMATIC CODE FIXING FUNCTION

def automatic_code_fixer(code, issues):

    prompt = f"""
    You are an expert Python software engineer.

    Your task is to fix the submitted Python code.

    ORIGINAL CODE:
    {code}

    DETECTED ISSUES:
    {issues}

    INSTRUCTIONS:

    1. Identify the actual errors.
    2. Fix confirmed errors.
    3. Make only necessary changes.
    4. Preserve the original program's intended behavior.
    5. Do not invent requirements.
    6. Return the complete corrected Python code.
    7. Put the corrected code inside one Python code block.
    8. Do not include explanations inside the code block.

    If an issue is only a possible warning, do not change
    program behavior just to remove that warning.
    """

    response = llm.invoke(prompt)

    ai_result = response.content

    corrected_code = extract_code(ai_result)

    # 3. VALIDATE CORRECTED CODE SYNTAX

    try:

        ast.parse(corrected_code)

        return {
            "status": "success",
            "message": "Corrected code passed syntax validation.",
            "corrected_code": corrected_code
        }

    except SyntaxError as error:

        return {
            "status": "validation_failed",
            "message": error.msg,
            "line": error.lineno,
            "corrected_code": corrected_code
        }


# 4. TEST AUTOMATIC CODE FIXING

buggy_code = """
a = 10

if a > 5
    print(a)
"""


# 5. DETECT BUGS

detected_issues = analyze_code(buggy_code)


print("=" * 60)
print("AUTOMATIC CODE FIXING ENGINE")
print("=" * 60)

print("\nORIGINAL CODE:")
print(buggy_code)

print("\nDETECTED ISSUES:")

for issue in detected_issues:
    print(issue)


# 6. GENERATE CORRECTED CODE

fix_result = automatic_code_fixer(
    buggy_code,
    detected_issues
)


# 7. DISPLAY FIXING RESULT

print("\n" + "=" * 60)
print("CODE FIXING RESULT")
print("=" * 60)

print("STATUS:", fix_result["status"])

print("MESSAGE:", fix_result["message"])

print("\nCORRECTED CODE:")
print(fix_result["corrected_code"])


# 8. FINAL VALIDATION

if fix_result["status"] == "success":

    print("\nSUCCESS!")
    print("Corrected code has valid Python syntax.")

else:

    print("\nWARNING!")
    print("AI-generated code did not pass syntax validation.")

print("\nAUTOMATIC CODE FIXING COMPLETED!")
````

````python
# ============================================================
# DAY 1 - FIRST AI API CALL + SWEET CRUMBS BAKERY CHATBOT
# ============================================================

!pip install -q openai

from google.colab import userdata
from openai import OpenAI

print("=" * 55)
print("DAY 1 - MY FIRST AI API CALL")
print("=" * 55)

# STEP 1: Read API key safely from Colab Secrets
api_key = userdata.get("GROQ_API_KEY")

if not api_key:
    raise ValueError(
        "GROQ_API_KEY not found. Check Colab Secrets."
    )

print("\nAPI key loaded successfully!")

# STEP 2: Connect to Groq
client = OpenAI(
    api_key=api_key,
    base_url="https://api.groq.com/openai/v1"
)

print("Groq client connected successfully!")

# STEP 3: Make our first simple API call
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    temperature=0.2,
    messages=[
        {
            "role": "system",
            "content": "You are a helpful AI assistant."
        },
        {
            "role": "user",
            "content": "Explain what an AI agent is in simple English."
        }
    ]
)

print("\nMY FIRST AI RESPONSE:")
print(response.choices[0].message.content)

# STEP 4: Display token usage
print("\nTOKEN USAGE:")
print("Input tokens:", response.usage.prompt_tokens)
print("Output tokens:", response.usage.completion_tokens)
print("Total tokens:", response.usage.total_tokens)


# ============================================================
# TODAY'S BUILD - SWEET CRUMBS HOME BAKERY
# ============================================================

SYSTEM = """
ROLE:
You are the friendly AI assistant for Sweet Crumbs Home Bakery.

FACTS:
1. We bake cakes after receiving orders.
2. Orders require 24 hours' notice.
3. All cakes have eggless versions.
4. Plum cake costs Rs 500.
5. Chocolate cake costs Rs 650.
6. Cupcakes cost Rs 80.

RULES:
1. Never promise a delivery time.
2. Do not invent prices or bakery policies.
3. If information is not available, say:
   "I'll check with the owner."
4. Keep answers under 50 words.
5. Reply in English.
"""

bakery_response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    temperature=0.2,
    messages=[
        {
            "role": "system",
            "content": SYSTEM
        },
        {
            "role": "user",
            "content": "Can I get an eggless cake today evening?"
        }
    ]
)

print("\n" + "=" * 55)
print("SWEET CRUMBS BAKERY CHATBOT")
print("=" * 55)

print("\nCUSTOMER QUESTION:")
print("Can I get an eggless cake today evening?")

print("\nAI CHATBOT ANSWER:")
print(bakery_response.choices[0].message.content)

print("\nTOKEN USAGE:")
print("Input tokens:", bakery_response.usage.prompt_tokens)
print("Output tokens:", bakery_response.usage.completion_tokens)
print("Total tokens:", bakery_response.usage.total_tokens)

print("\nDAY 1 PRACTICAL TASK COMPLETED!")
````
